### Multithreading

````python
from threading import Thread
from time import sleep
def Display1():
    for i in range(10):
        sleep(0.5)
        print("Display",i+1)
def Display2():
    for i in range(10):
        sleep(0.3)
        print("print",i+1)
t1 = Thread(target = Display1)
t2 = Thread(target = Display2)
t1.start()
t2.start()
t1.join()
t2.join()
print("END")
````

````python
class A(Thread):
    def run(self):
        print("Class A run method")

class B(Thread):
    def run(self):
        print("Class B run method")

t1 = A()
t2 = B()

t1.start()
t2.start()
````

### execute()

````python
import sqlite3
conn = sqlite3.connect("student.db")
cur = conn.cursor()
cur.execute('INSERT into course1 VALUES(103,"python");')
cur.execute('INSERT into course1 VALUES(104,"Java");')
conn.commit()
conn.close()
````

### executemany()

````python
import sqlite3
conn = sqlite3.connect("student.db")
cur = conn.cursor()
data = [(105,"c"),(106,"c++")]
cur.executemany('INSERT into course1 VALUES(?,?);',data)
conn.commit()
conn.close()
````

### executescript()

````python
import sqlite3
conn = sqlite3.connect("student.db")
cur = conn.cursor()
s = '''
update course1 set name = "Data science using python" where id=103;
delete from course1 where id=106;
'''
cur.executescript(s)
conn.commit()
conn.close()
````

### Iterating through records without fetch methods:

````python
import sqlite3
conn = sqlite3.connect("student.db")
cur = conn.cursor()
cur.execute("select * from course1;")
for row in cur:
    print(row)
conn.commit()
conn.close()
````

### fetchone():

````python
import sqlite3
conn = sqlite3.connect("student.db")
cur = conn.cursor()
cur.execute("select * from course1;")

row = cur.fetchone()
row = cur.fetchone()
print(row)

conn.close()
````

### fetchmany()

````python
import sqlite3
conn = sqlite3.connect("student.db")
cur = conn.cursor()
cur.execute("select * from course1;")

row = cur.fetchmany(2)

print(row)

conn.close()
````

````python
import sqlite3
conn = sqlite3.connect("student.db")
cur = conn.cursor()
cur.execute("select * from course1;")

row = cur.fetchall()

print(row)

conn.close()
````

### count the no.of records in the table

````python
import sqlite3
conn = sqlite3.connect("student.db")
cur = conn.cursor()
cur.execute("select count(*) from course1;")

row = cur.fetchone()

print("Count of records:",row[0])

conn.close()
````
