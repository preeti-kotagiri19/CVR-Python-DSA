List Comprehension is a concise way to create lists in Python using a single line of code.

### Leetcode problems
1. Fizz Buzz(412)
2. Squares of a Sorted Array (#977)
3. Find Numbers with Even Number of Digits (#1295)
4. Product of Array Except Self (#238)

````python
#Squares of numbers
#input : [1,2,3,4,5] output : [1,4,9,16,25]
li = list(map(int,input().split()))
res = []
for i in li:
    res.append(i ** 2)
print(res)
````

````python
li = list(map(int,input().split()))
res = [i ** 2 for i in li]
print(res)
````

````python
#Even numbers from list
#input:[1,2,3,4,5] output:[2,4]
li = list(map(int,input().split()))
res = []
for ele in li:
    if ele % 2 == 0:
        res.append(ele)
print(res)
````

````python
li = list(map(int,input().split()))
res = [ele for ele in li if ele % 2 == 0]
print(res)
````

### Flatten a 2D list
```
input : [[1,2],[3,4],[5,6]]
output : [1,2,3,4,5,6]
```

````python
matrix = [[1,2],[3,4],[5,6]]
res = []
for i in matrix:
    for j in i:
        res.append(j)
print(res)
````

````python
matrix = [[1,2],[3,4],[5,6]]
res = [j for i in matrix for j in i]
print(res)
````

````python
#Remove negative numbers
#input : [-4,5,-2,63,78,-100]  output : [5,63,78]
li = [-4,5,-2,63,78,-100]
res = [ele for ele in li if ele >= 0]
print(res)
````

### 1295. Find Numbers with Even Number of Digits

````python
nums = [12,345,2,6,7896]
count = 0
for n in nums:
    if len(str(n)) % 2 == 0:
        count += 1
print(count) 
````

````python
nums = [12,345,2,6,7896]
print(len([n for n in nums if len(str(n)) % 2 == 0]))
````

### 412. Fizz Buzz

````python
n = int(input())
answer = []
for i in range(1,n+1):
    if i % 3 == 0 and i % 5 == 0:
        answer.append("fizzbuzz")
    elif i % 3 == 0:
        answer.append("fizz")
    elif i % 5 == 0:
        answer.append("buzz")
    else:
        answer.append(str(i))
print(answer)
````

````python
n = int(input())
answer = ["fizzbuzz" if i % 3 == 0 and i % 5 == 0
          else "fizz" if i % 3 == 0
          else "buzz" if i % 5 == 0
          else str(i) for i in range(1,n+1)]
print(answer)
````

### Lambda functions
A Lambda Function is an anonymous function written in one line

````python
obj = lambda x : x * x
````

````python
obj(5)
````

````python
def fun(x):
    return x * x
````

````python
li = [1,2,3,4,5]
res = []
for i in li:
    res.append(fun(i))
print(res)
````

````python
list(map(fun,li))
````

````python
list(map(lambda x : x * x,li))
````

````python
#li = [1,2,3,4,5] output : [2,4]
````

````python
def even(x):
    if x % 2 == 0:
        return True
    else:
        return False
````

````python
li = [1,2,3,4,5]
list(map(even,li))
````

````python
list(filter(even,li))
````

````python
li = [1,2,3,4,5]
list(filter(lambda x : x % 2 == 0,li))
````

````python
li = ['ab','mam','python','madam','wxyz']
#['python','madam','wxyz','mam','ab']
sorted(li,key=lambda s : len(s),reverse = True)
````

````python
students = [('a',60),('b',35),('c',90),('d',70)]
sorted(students,key=lambda x : x[1],reverse = True)
````
