### List vs Linked list:
#### Python List
##### Advantages:
1. Fast indexing: O(1)
2. Easy to use
3. Built-in support

##### Disadvantages:
1. Insertion/deletion at front is O(n)

#### Linked List
##### Advantages:
1. O(1) insertion/deletion at the head
2. Dynamic size

Cons

O(n) access
Extra memory for pointers
Must be implemented manually in Python

### Linked list:
1. singly LL
2. Doubly LL
3. Circular LL

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    def Insert_at_begin(self,val):
        new_node = Node(val)
        new_node.next = self.head
        self.head = new_node
        
    def Insert_at_end(self,val):
        new_node = Node(val)
        if not self.head:
            self.head = new_node
            return
        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = new_node
    
    def Insert_at_position(self,val,pos):
        new_node = Node(val)
        if not self.head:
            self.head = new_node
            return
        temp = self.head
        for _ in range(pos-2):
            if temp.next is None:
                print("Invalid position")
                return 
            temp = temp.next
        if temp.next is None:
                print("Invalid position")
                return
        new_node.next = temp.next
        temp.next = new_node
    
    def delete_at_begin(self):
        if self.head:
            self.head = self.head.next

    def delete_at_end(self):
        if not self.head:
            return
        if not self.head.next:
            self.head = None
            return
        temp = self.head
        while temp.next.next:
            temp = temp.next
        temp.next = None
        
    def delete_with_value(self,val):
        if self.head is None:
            return
        if self.head.data == val:
            self.head = self.head.next
        temp = self.head
        while temp.next and temp.next.data != val:
            temp = temp.next
        if temp.next:
            temp.next = temp.next.next

    def delete_node_at_position(self,pos):
        if self.head is None:
            return
        if pos == 1:
            self.head = self.head.next
            return
        temp = self.head
        for _ in range(pos - 2):
            if temp is None or temp.next is None:
                print("Invalid Position")
                return
            temp = temp.next
        if temp.next is None:
            print("Invalid Position")
            return
        temp.next = temp.next.next

    def Display(self):
        temp = self.head
        while not temp.next is None:
            print(temp.data)
            temp = temp.next
        print(temp.data)
        
    def Display_revised(self):
        temp = self.head
        while temp:
            print(temp.data,end="==>")
            temp = temp.next 
        print()
            
    def Count_nodes(self):
        temp = self.head
        count = 0
        while temp:
            count += 1
            temp = temp.next 
        return count

ll = LinkedList()
ll.Insert_at_begin(10)
ll.Insert_at_begin(20)
ll.Insert_at_begin(30)
ll.Insert_at_end(40)
ll.Insert_at_end(50)
ll.Insert_at_position(60,3)
#ll.Display()
ll.Display_revised()
#print(ll.Count_nodes())
ll.delete_at_begin()
ll.Display_revised()
ll.delete_at_end()
ll.Display_revised()
ll.delete_with_value(60)
ll.Display_revised()
ll.delete_node_at_position(2)
ll.Display_revised()
```
