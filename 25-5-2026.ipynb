### Operator:
```
1. Arithmetic operators ==> +,-,*,/,%(modulus),//(floor division),**(exponent)
2. Relational operators ==> <,<=,>,>=,!=,==
3. Logical operators ==> and,or,not
4. Assignment operators ==> +=,-=,*=,/=,%=,//=,**=
5. Bit-wise operators ==> &,|,^(XOR),~,<<,>>
6. Membership operators ==> in,not in
7. Identity operators ==> is,is not
```

````python
x = 13      # 0000 1101
y = 6       # 0000 0110
print(x & y)# 0000 0100 ==> 4
print(x | y)# 0000 1111 ==> 15
print(x ^ y)# 0000 1011 ==> 11
print(x << 2)# 0011 0100
print(x >> 2)# 0000 0011
print(~x) #==> -1 * (x+1)
````
