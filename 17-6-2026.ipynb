```python
def fib(n):
    if n <= 1:
        return n
    return fib(n-2) + fib(n-1)

n = int(input())
print(fib(n))
```

### Memoization:

```python
def fib(n):
    memo = {}
    if n <= 1:
        return n
    if n in memo:
        return memo[n]
    memo[n] = fib(n-2) + fib(n-1)
    return memo[n]

n = int(input())
print(fib(n))
```

### Iterative approach:

```python
def fib(n):
    a,b = 0,1
    for i in range(2,n+1):
        a,b = b,a+b
    return b

n = int(input())
print(fib(n))
```

### Tabulation:

```python
def fib(n):
    dp = [0]*(n+1)
    dp[0] = 0
    dp[1] = 1
    for i in range(2,n+1):
        dp[i]= dp[i-2] + dp[i-1]
    return dp[n]

n = int(input())
print(fib(n))
```

### 70. Climbing Stairs

```python
class Solution:
    def climbStairs(self, n: int) -> int:
        if n <= 2:
            return n
        dp = [0] * (n+1)
        dp[1] = 1
        dp[2] = 2
        for i in range(3,n+1):
            dp[i] = dp[i-2] + dp[i-1]
        return dp[n]
```

### 3693. Climbing Stairs II
#### Solution-1

```python
from typing import List
class Solution:
    def climbStairs(self, n: int, costs: List[int]) -> int:
        dp = [float('inf')] * (n + 1)
        dp[0] = 0
        for i in range(1, n + 1):
            for jump in range(1, 4):
                prev = i - jump
                if prev >= 0:
                    dp[i] = min(dp[i],dp[prev] + costs[i - 1] + jump * jump)

        return dp[n]
        
```

#### Solution-2

```python
class Solution:
    def climbStairs(self, n: int, costs: List[int]) -> int:
        f0 = f1 = f2 = 0
        for cost in costs:
            f0, f1, f2 = f1, f2, min(f0 + 9, f1 + 4, f2 + 1) + cost
        return f2
```

### 198. House Robber
#### Solution-1

```python
class Solution:
    def rob(self, nums: List[int]) -> int:
        n = len(nums)
        if n == 1:
            return nums[0]

        dp = [0] * n

        dp[0] = nums[0]
        dp[1] = max(nums[0], nums[1])

        for i in range(2, n):
            dp[i] = max(dp[i-1], nums[i] + dp[i-2])

        return dp[-1]
```

#### Solution-2

```python
class Solution:
    def rob(self, nums: List[int]) -> int:
        prev2 = 0
        prev1 = 0
        for ele in nums:
            cur = max(prev1,ele + prev2)
            prev2 = prev1
            prev1 = cur
        return prev1
```

### 1143. Longest Common Subsequence

```python
class Solution:
    def longestCommonSubsequence(self, text1: str, text2: str) -> int:
        m = len(text1)
        n = len(text2)
        dp = [[0]*(n+1) for _ in range(m+1)]
        for i in range(1, m+1):
            for j in range(1, n+1):
                if text1[i-1] == text2[j-1]:
                    dp[i][j] = 1 + dp[i-1][j-1]
                else:
                    dp[i][j] = max(dp[i-1][j],dp[i][j-1])

        return dp[m][n]
```
