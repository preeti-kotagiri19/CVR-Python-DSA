### Implement Trie (Prefix Tree) - 208

````python
class TrieNode:
    def __init__(self):
        self.children = {}
        self.is_end_of_word = False

class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self,string):
        curr = self.root
        for ch in string:
            if ch not in curr.children:
                curr.children[ch] = TrieNode()
            curr = curr.children[ch]
        curr.is_end_of_word = True

    def search(self,string):
        curr = self.root
        for ch in string:
            if ch not in curr.children:
                return False
            curr = curr.children[ch]
        return curr.is_end_of_word   

    def startsWith(self, prefix):
        current = self.root
        for char in prefix:
            if char not in current.children:
                return False
            current = current.children[char]
        return True
````

### Replace Words - 648

````python
from typing import List
class TrieNode:
    def __init__(self):
        self.children = {}
        self.is_end_of_word = False

class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self,string):
        curr = self.root
        for ch in string:
            if ch not in curr.children:
                curr.children[ch] = TrieNode()
            curr = curr.children[ch]
        curr.is_end_of_word = True

    def find_root(self,string):
        curr = self.root
        pref = ""
        for ch in string:
            if ch not in curr.children:
                return string
            curr = curr.children[ch]
            pref += ch
            if curr.is_end_of_word:
                return pref
        return string

class Solution:
    def replaceWords(self, dictionary: List[str], sentence: str) -> str:
        t = Trie()
        for root_word in dictionary:
            t.insert(root_word)
        words = sentence.split()
        res = []
        for word in words:
            res.append(t.find_root(word))
        return " ".join(res)
        
````
