### 1672. Richest Customer Wealth

````python
#Solution-1
accounts = [[1,5],[7,3],[3,5]]
max(list(map(lambda x:sum(x),accounts)))
````

````python
#Solution-2
res = []
for i in accounts:
    res.append(reduce(lambda a,b:a+b,i))
print(max(res))
````

````python
#Solution-3
max([reduce(lambda a,b:a+b,i) for i in accounts])
````

### 905. Sort Array By Parity

````python
nums = [3,1,2,4]
res = []
for ele in nums:
    if ele % 2 == 0:
        res.insert(0,ele)
    else:
        res.append(ele)
print(res)
````

````python
sorted(nums,key=lambda x : x%2)
````

````python
class A:
    def __init__(self,x):
        self.x = x
    def display(self):
        print(self.x)

obj1 = A(10)
obj2 = A(20)

obj1.display()
obj2.display()
````

````python
from math import pi
class Circle:
    def __init__(self,radius):
        self.radius = radius
    def Area(self):
        print(pi * self.radius * self.radius)
    def Perimeter(self):
        print(2 * pi * self.radius)

c = Circle(5)
c.Area()
c.Perimeter()
        
````

### Inheritance

````python
class A:
    def display2(self):
        print("class A display method")
class B:
    def display3(self):
        print("class B display method")
class C(A,B):
    def display1(self):
        print("class C display method")

c = C()

c.display()
````

### Polymorphism:
1. Compile-time polymorphism
    1. Function overloading
    2. operator overloading
   
2. Run-time polymorphism
    1. Method overriding

````python
'''
def add(x,y):
    return x+y
def add(x,y,z):
    return x+y+z
def add(x,y,z,a):
    return x+y+z+z
'''
def add(*args):
    return sum(args)
    
print(add(10,20))
print(add(10,20,30))
print(add(10,20,30,40))
````

````python
class A:
    def __init__(self,val):
        self.val = val
        
    def __add__(self,x):
        return self.val + x.val
        
a = A(10)
b = A(20)
print(a + b)
print(a - b)
````

````python
#Method overriding
class A:
    def display(self):
        print("class A display method")
class B(A):
    def display(self):
        print("class B display method")
b = B()
b.display()
````
