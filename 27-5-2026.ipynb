### Recursion:
Calling a function from the same function definition

1. Base case
2. Recursive case

````python
#sum of n natural numbers
def Natural(n):
    s = 0
    for i in range(n,0,-1):
        s += i
    return s
    
print(Natural(5))
print(Natural(10))
````

````python
def Natural_recursion(n):
    if n == 1:
        return 1
    return n + Natural_recursion(n-1)

print(Natural_recursion(5))
print(Natural_recursion(10))
````

````python
#Factorial of a number
def Factorial(n):
    if n < 0:
        return "No factorial for -ve"
    elif n == 0 or n == 1:
        return 1
    else:
        f = 1
        for i in range(n,1,-1):
            f *= i
        return f

print(Factorial(5))#120
print(Factorial(3))#6
````

````python
def Factorial_recursion(n):
    if n < 0:
        return "No factorial for -ve"
    elif n == 0 or n == 1:
        return 1
    return n * Factorial_recursion(n-1)

print(Factorial_recursion(5))
print(Factorial_recursion(3))
````

````python
#nth Fibonacci series number
def Fibonacci(n):
    if n == 0 or n == 1:
        return 0
    elif n == 2:
        return 1
    else:
        a,b = 0,1
        for i in range(2,n):
            a,b = b,a+b
        return b  
print(Fibonacci(2))#1
print(Fibonacci(5))#3
print(Fibonacci(7))#8
````

````python
def Fibonacci_recursion(n):
    if n == 0 or n == 1:
        return 0
    elif n == 2:
        return 1
    return Fibonacci_recursion(n-2) + Fibonacci_recursion(n-1)

print(Fibonacci_recursion(5))
````

````python
def Array_sum(li):
    s = 0
    for i in range(len(li)):
        s += li[i]
    return s
print(Array_sum([1,2,3,4,5]))#15
````

````python
def Array_sum_rec(li):
    if len(li) == 0:
        return 0
    return li[-1] + Array_sum_rec(li[:-1])
    
print(Array_sum([1,2,3,4,5]))#15
````

````python
#Reverse an array(inplace)
def Reverse_array(arr):
    n = len(arr)
    for i in range(n//2 + 1):
        arr[i],arr[n-1-i] = arr[n-1-i],arr[i]
    return arr

print(Reverse_array([1,2,3,4,5]))#[5,4,3,2,1]
````

````python
def Reverse_array_rec(arr,left,right):
    if left >= right:
        return
    arr[left],arr[right] = arr[right],arr[left]
    Reverse_array_rec(arr,left+1,right-1)
    return arr

li = [1,2,3,4,5]
print(Reverse_array_rec(li,0,4))
````

````python
#String reversing
def String_Rev(s):
    if len(s) == 0:
        return ""
    return s[-1] + String_Rev(s[:-1])
    
def Check_palindrome(s):
    if s == String_Rev(s):
        return "palindrome"
    else:
        return "not palindrome"
    
print(Check_palindrome("abc"))
````
