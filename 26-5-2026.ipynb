### Question:
```
1. Accept 3 subjects marks
2. check whether all the subjects marks are >= 35 or not
3. if all subjects marks >=35 Then calculate avg and give grades
4. avg >= 75 ==> 'A'
   avg >= 60 and avg < 75 ==> 'B'
   avg >= 35 and avg < 60 ==> 'C'
   avg < 35 ==> 'Fail'
```

````python
m1,m2,m3 = map(int,input().split())
if m1 >= 35 and m2 >= 35 and m3 >= 35:
    avg = (m1 + m2 + m3)/3
    if avg >= 75:
        print('A')
    elif avg >= 60 and avg < 75:
        print('B')
    elif avg >= 35 and avg < 60:
        print('C')
else:
    print("Fail")
````

```
input : 156

input : 23894
```

````python
n = int(input())
s = 0
while n > 0:
    s += (n % 10)
    n = n // 10
print(s)
````

````python
n = int(input())
for i in range(1,n+1):
    for j in range(i):
        print("* ",end=" ")
    print()
````

```
Hollow Square
n = 4
Output:
* * * *
*     *
*     *
* * * *
```

````python
n = int(input())
for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            print("*",end="")
        else:
            print(" ",end="")
    print()
````
