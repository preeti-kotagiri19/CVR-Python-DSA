```python
from threading import Thread
lock = threading.Lock()
balance = 1000
def update(amount):
    global balance
    lock.acquire()
    balance += amount
    lock.release()
    print(balance)

t1 = Thread(target=update,args=(500,))
t2 = Thread(target=update,args=(200,))
t1.start()
t2.start()
```

### 643. Maximum Average Subarray I

### Brute-force approach

```python
class Solution:
    def findMaxAverage(self, nums: List[int], k: int) -> float:
        n = len(nums)
        res = []
        for i in range(n-k+1):
            res.append(sum(nums[i:k+i])/k)
        print(max(res))
```

```python
obj = Solution()
obj.findMaxAverage([1,12,-5,-6,50,3],4)
```

### optimal solution 

```python
class Solution:
    def findMaxAverage(self, nums: List[int], k: int) -> float:
        w_sum = sum(nums[:k])
        max_sum = w_sum
        for i in range(k,len(nums)):
            w_sum = w_sum - nums[i-k] + nums[i]
            if w_sum > max_sum:
                max_sum = w_sum
        print(max_sum / k)

obj = Solution()
obj.findMaxAverage([1,12,-5,-6,50,3],4)
```

### 209. Minimum Size Subarray Sum

### Brute-force approach

```python
class Solution:
    def minSubArrayLen(self, target: int, nums: list[int]) -> int:
        min_len = float('inf')
        for i in range(len(nums)):
            s = 0
            for j in range(i,len(nums)):
                s += nums[j]
                if s >= target:
                    min_len = min(min_len,j-i+1)
                    break
        return min_len if min_len != float('inf') else 0
                
obj = Solution()
obj.minSubArrayLen(7,[2,3,1,2,4,3])
```
