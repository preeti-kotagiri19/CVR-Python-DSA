### Abstraction:
Hiding implementation details from the user

````python
from abc import ABC,abstractmethod
````

````python
class Bank(ABC):
    @abstractmethod
    def transaction(self):
        pass
````

````python
class HDFC(Bank):
    def transaction(self):
        print("HDFC transactions history")
````

````python
h = HDFC()
h.transaction()
````

### File handling:
1. open(file_name,mode) ==> file object
2. close()

### Types of modes:
```
1. 'r' ==> read
   1. file does not exist ==> error
   2. file exist ==> reading can happen
2. 'w' ==> write
   1. file does not exist ==> new file created
   2. file exist with some text ==> old text is removed and new text will be added
   3. file exist with no text ==> text is added as a new text
3. 'a' ==> append
   1. file does not exist ==> new file created
   2. file exist with some text ==> new text will be added at the end of old
   3. file exist with no text ==> text is added as a new text
4. 'x' ==> new file
   1. file does not exist ==> creates a new file
   2. file exist ==> error
```

````python
#creating a file and adding text to the file
f = open("demo.txt","w")
st = '''Line-1
Line-2
Line-3'''
f.write(st)
f.close()
````

````python
#Reading an existing file
f = open("demo.txt","r")
data = f.read()
print(data)
f.close()
````

````python
#Convert File Content to Uppercase
with open("demo.txt","r") as f:
    data = f.read()
print("Before converting:",data)

with open("demo.txt","w") as f:
    f.write(data.upper())
with open("demo.txt","r") as f:
    data = f.read()

print("After converting:",data)
````

````python
#Copy File Content from one file to another
with open("demo.txt","r") as f:
    data = f.read()
with open("sample.txt","w") as f1:
    f1.write(data)
with open("sample.txt","r") as f:
    d = f.read()
    print(d)
    
````

````python
#Count Vowels
with open("demo.txt","r") as f:
    data = f.read()
count = 0
for ch in data:
    if ch in "aeiouAEIOU":
        count += 1
print(count)
````

````python
#Frequency of Each Character
s = "preeti"
d = {}
for ch in s:
    if ch not in d:
        d[ch] = 1
    else:
        d[ch] += 1
print(d)
````

````python
d.get('z',0)
````

````python
s = "preeti"
d = {}
for ch in s:
    d[ch] = d.get(ch,0) + 1
print(d)
````

````python
##Frequency of Each Character from file text
with open("demo.txt","r") as f:
    s = f.read()
d = {}
for ch in s:
    d[ch] = d.get(ch,0) + 1
print(d)
````
