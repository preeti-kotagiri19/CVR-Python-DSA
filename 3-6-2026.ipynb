### Age verification

```python
class InvalidAgeError(Exception):
    pass
try:
    age = int(input())
    if age < 18:
        raise InvalidAgeError("Given age is <18")
    print("Eligible for voting")
    
except InvalidAgeError as arg:
    print("InvalidAgeError:",arg)
```

### password verification:
1. length should be >= 8
2. atleast one upper case character and one digit

```python
class PasswordIncorrectError(Exception):
    pass
try:
    password = input()
    if len(password) < 8:
        raise PasswordIncorrectError("Password length must be >= 8")
    else:
        if not any(ch.isupper() for ch in password):
            raise PasswordIncorrectError("Password must contain atleast one upper case character")
        if not any(ch.isdigit() for ch in password):
            raise PasswordIncorrectError("Password must contain atleast one digit")
    print("Password Accepted")
except PasswordIncorrectError as e:
    print(e)
        
```

### ATM withdrawl system

```python
class BalanceInsufficientError(Exception):
    pass

try:
    balance = int(input())
    withdrawl = int(input())
    if withdrawl > balance:
        raise BalanceInsufficientError("Withdrawl amount is > balance")
    balance -= withdrawl
    print("Current balance:",balance)
    
except BalanceInsufficientError as a:
    print(a)
```
